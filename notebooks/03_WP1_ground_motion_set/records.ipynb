```python
from pathlib import Path
import h5py
```

```python
def explore_h5(name, obj):
    """Recursive callback to print structure and metadata."""
    indent = "  " * name.count('/')
    if isinstance(obj, h5py.Group):
        print(f"{indent}Group: {name}")
    elif isinstance(obj, h5py.Dataset):
        print(f"{indent}Dataset: {name} (Shape: {obj.shape}, Dtype: {obj.dtype})")
    
    # Print attributes (metadata)
    for key, val in obj.attrs.items():
        print(f"{indent}    Attr -> {key}: {val}")
```

```python
fp = Path(r"C:\Users\clemettn\Documents\phd\data_raw\gm_records_for_analysis\EMSC-20160824_0000006__AQG__0.h5")
with h5py.File(fp, "r") as f:
    # 1. Map the entire hierarchy
    print("--- File Structure ---")
    f.visititems(explore_h5)
```
