```python
import json
from pathlib import Path

import numpy as np

from standes.groundmotion import response_spectrum
```

```python
record_path = Path(r"E:\gm_records_p695\fema_p695_120111.json")

with open(record_path, "r") as f:
    record = json.load(f)

acc = record["record"]
dt = record["dt"]

periods = np.round(np.linspace(0, 3, 10), 3)

sf = 1.235431471
sa = 0.73205         # g
```

```python
pga = np.abs(acc).max()

# compute the response spectrum for all periods except PGA
sa_ts, *_ = response_spectrum(acc, dt, periods[1:])
sa_ts = np.insert(sa_ts, 0, pga)
```

```python
# compute the average SA
avg_sa = np.prod(sa_ts * 10) ** (1 / len(sa_ts)) / 10

# the final converted value is the average SA multiplied by the scale factor
print(avg_sa)

```
